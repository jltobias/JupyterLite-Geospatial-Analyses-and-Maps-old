```python
from pathlib import Path
import sys
for p in [Path.cwd(), *Path.cwd().parents]:
    if (p / 'spatial_lite.py').exists():
        sys.path.insert(0, str(p))
        break
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import spatial_lite as sl

pd.set_option('display.max_columns', 20)
```

```python
df = sl.load_dataset('bike_share_trips.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
trips=df.copy()
trips['trip_km']=sl.haversine(trips.start_lon,trips.start_lat,trips.end_lon,trips.end_lat)
display(trips[['id','duration_min','trip_km','hour']].head())
sl.plot_points(trips.rename(columns={'start_lon':'lon','start_lat':'lat'}), color='duration_min', title='cartographic: bike trip starts')
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
numeric = df.select_dtypes(include='number')
corr = numeric.corr().round(2)
display(corr)
first = numeric.columns[0]
numeric[first].hist(bins=15); plt.title(f'Distribution of {first}'); plt.xlabel(first); plt.ylabel('count'); plt.show()
```
