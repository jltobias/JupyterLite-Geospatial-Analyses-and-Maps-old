```python
df = sl.load_dataset('iris_geo.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
work = df.copy()
work['class'] = sl.classify_quantiles(work['petal_length'], 5)
display(work.groupby('class')['petal_length'].agg(['count','mean','min','max']).round(2))
sl.plot_points(work, color='petal_length', title='machine-learning: petal_length by location')
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
iris = sl.load_dataset('iris_geo.csv')
iris['cluster'] = sl.simple_kmeans(iris, ['sepal_length','sepal_width','petal_length','petal_width'], k=3)
display(pd.crosstab(iris['species'], iris['cluster']))
sl.plot_points(iris, color='cluster', title='Unsupervised clusters from Iris features')
```
