```python
df = sl.load_dataset('meuse.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
work = df.copy()
work['class'] = sl.classify_quantiles(work['zinc'], 5)
display(work.groupby('class')['zinc'].agg(['count','mean','min','max']).round(2))
sl.plot_points(work, color='zinc', title='kriging: zinc by location')
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
pts = sl.load_dataset('meuse.csv')
grid = sl.load_dataset('landcover_grid.csv').sample(120, random_state=7).copy()
grid['zinc_idw'] = sl.idw(pts, 'zinc', grid.lon, grid.lat, power=2)
display(grid[['cell_id','zinc_idw']].head().round(1))
sl.plot_points(grid, color='zinc_idw', title='IDW interpolation: predicted zinc')
```
