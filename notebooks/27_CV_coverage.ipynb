```python
df = sl.load_dataset('nyc_taxi_sample.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
trips = df.copy()
trips['speed_kmh'] = trips['distance_km'] / (trips['duration_min']/60)
display(trips.groupby(['pickup_zone','dropoff_zone'])[['distance_km','fare_usd','duration_min']].mean().round(2).head(12))
# pickup geography as a map-like scatter
sl.plot_points(trips.rename(columns={'pickup_lon':'lon','pickup_lat':'lat'}), color='fare_usd', title='coverage: taxi pickup geography')
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
events = sl.load_dataset('earthquake_points.csv')
center_lon, center_lat = events.lon.mean(), events.lat.mean()
events['distance_to_center_km'] = sl.haversine(center_lon, center_lat, events.lon, events.lat)
display(events.nsmallest(10, 'distance_to_center_km')[['id','magnitude','distance_to_center_km']].round(2))
events.distance_to_center_km.hist(bins=18); plt.title('Distance decay / proximity pattern'); plt.xlabel('km'); plt.ylabel('count'); plt.show()
```
