```python
from pathlib import Path
import sys
for p in [Path.cwd(), *Path.cwd().parents]:
    if (p / 'spatial_lite.py').exists():
        sys.path.insert(0, str(p))
        break
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import spatial_lite as sl

pd.set_option('display.max_columns', 20)
```

```python
df = sl.load_dataset('landcover_grid.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
grid = df.copy()
sl.plot_grid(grid, 'elevation_m', 'suitability: elevation surface')
sl.plot_grid(grid, 'ndvi', 'suitability: vegetation index')
summary = grid.groupby('landcover')[['elevation_m','ndvi']].mean().round(3)
display(summary)
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
sites=sl.load_dataset('retail_sites.csv').copy()
sites['suitability'] = (sites.sales_index.rank(pct=True)*0.55 + sites.parking_spaces.rank(pct=True)*0.25 + sites.floor_area.rank(pct=True)*0.20)
display(sites.nlargest(10,'suitability')[['id','sales_index','parking_spaces','floor_area','suitability']].round(2))
sl.plot_points(sites, color='suitability', title='Candidate site suitability')
```
