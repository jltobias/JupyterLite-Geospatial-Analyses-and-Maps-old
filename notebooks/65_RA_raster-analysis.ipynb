```python
from pathlib import Path
import sys
for p in [Path.cwd(), *Path.cwd().parents]:
    if (p / 'spatial_lite.py').exists():
        sys.path.insert(0, str(p))
        break
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import spatial_lite as sl

pd.set_option('display.max_columns', 20)
```

```python
df = sl.load_dataset('landcover_grid.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
grid = df.copy()
sl.plot_grid(grid, 'elevation_m', 'raster-analysis: elevation surface')
sl.plot_grid(grid, 'ndvi', 'raster-analysis: vegetation index')
summary = grid.groupby('landcover')[['elevation_m','ndvi']].mean().round(3)
display(summary)
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
grid = df.copy()
grid['slope_proxy'] = grid.groupby('y')['elevation_m'].diff().abs().fillna(0) + grid.groupby('x')['elevation_m'].diff().abs().fillna(0)
display(grid[['cell_id','elevation_m','ndvi','slope_proxy','landcover']].head())
sl.plot_grid(grid, 'slope_proxy', 'Derived raster: slope proxy')
```
