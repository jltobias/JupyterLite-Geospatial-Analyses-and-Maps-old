```python
from pathlib import Path
import sys
for p in [Path.cwd(), *Path.cwd().parents]:
    if (p / 'spatial_lite.py').exists():
        sys.path.insert(0, str(p))
        break
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import spatial_lite as sl

pd.set_option('display.max_columns', 20)
```

```python
df = sl.load_dataset('iris_geo.csv')
sl.quick_summary(df)
```

## 1. See the spatial pattern

Before running an algorithm, inspect scale, outliers, and geography. This lowers cognitive load and makes later transformations easier to diagnose.

```python
work = df.copy()
work['class'] = sl.classify_quantiles(work['petal_length'], 5)
display(work.groupby('class')['petal_length'].agg(['count','mean','min','max']).round(2))
sl.plot_points(work, color='petal_length', title='spatial-join: petal_length by location')
```

## 2. Apply the analysis idea

The code below uses small, transparent calculations that run in Pyodide without compiled geospatial dependencies.

```python
left = sl.load_dataset('retail_sites.csv').head(12)
right = sl.load_dataset('weather_stations.csv')
joined = sl.nearest_join(left, right)
display(joined[['id','nearest_id','distance_km']].round(2))
```
